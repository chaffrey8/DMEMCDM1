```python
import time
import random
import matplotlib.pyplot as plt

# -----------------------------------------
# Algoritmo cuadrático: Bubble Sort (O(n²))
# -----------------------------------------
def bubble_sort(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n):
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a


# -----------------------------------------
# Función para medir el tiempo de ejecución
# -----------------------------------------
def medir_tiempo(func, n):
    vec = [random.randint(0, 10**6) for _ in range(n)]
    inicio = time.time()
    func(vec)
    fin = time.time()
    return fin - inicio


# -----------------------------------------
# Tamaños a evaluar
# -----------------------------------------
tamanios = [2000, 10000, 15000, 20000]

tiempos_sorted = []
tiempos_bubble = []

for n in tamanios:
    tiempos_sorted.append(medir_tiempo(sorted, n))      # O(n log n)
    tiempos_bubble.append(medir_tiempo(bubble_sort, n)) # O(n²)


# -----------------------------------------
# Gráfica de comparación
# -----------------------------------------
plt.figure()
plt.plot(tamanios, tiempos_sorted, marker="o", label="sorted()")
plt.plot(tamanios, tiempos_bubble, marker="o", label="bubble_sort")
plt.xlabel("Tamaño del vector")
plt.ylabel("Tiempo (s)")
plt.title("Comparación de tiempos: O(n log n) vs O(n²)")
plt.legend()
plt.tight_layout()
plt.show()
```
